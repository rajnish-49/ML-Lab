```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import KBinsDiscretizer
from sklearn.naive_bayes import CategoricalNB
from sklearn.metrics import accuracy_score

# Load dataset
df = pd.read_csv("student_dataANN.csv")

# Create target
conditions = [
    df["Final_Score"] < 60,
    (df["Final_Score"] >= 60) & (df["Final_Score"] < 75),
    df["Final_Score"] >= 75
]
choices = [0, 1, 2]
df["PerformanceClass"] = np.select(conditions, choices)

# Features and target
X = df[["Study_Hours", "Attendance", "Previous_Score"]]
y = df["PerformanceClass"]

# Discretize features (like your BN bins)
kbd = KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='quantile')
X_binned = kbd.fit_transform(X)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X_binned, y, test_size=0.2, random_state=42, stratify=y
)

# Model
model = CategoricalNB()
model.fit(X_train, y_train)

# Test
y_pred = model.predict(X_test)
print("Test Accuracy:", accuracy_score(y_test, y_pred))

# Prediction
input_data = [[6, 75, 70]]
input_binned = kbd.transform(input_data)

pred = model.predict(input_binned)[0]
probs = model.predict_proba(input_binned)[0]

labels = ["Low", "Medium", "High"]

print("Predicted Class:", labels[pred])
print("Probabilities:", probs)
```
